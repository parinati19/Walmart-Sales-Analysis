# 10. Store-wise Average Weekly Sales Analysis

## Objective

To calculate the average weekly sales of each Walmart store and identify the best-performing stores based on average sales.

# 10. Store-wise Average Weekly Sales Analysis

## Objective
To calculate the average weekly sales of each Walmart store and identify the best-performing stores based on average sales.

## Methodology
- Grouped the dataset by Store.
- Calculated the average weekly sales using the `mean()` function.
- Sorted the stores in descending order based on average weekly sales.
- Visualized the top 10 stores using a bar chart.

## Observation
- Store 20 has the highest average weekly sales.
- Store 4 ranks second in average weekly sales.
- Stores 14, 13, 2, and 10 also perform consistently well.

# 14. Store-wise Sales Consistency

## Objective

To analyse the consistency of weekly sales for each Walmart store using Standard Deviation.

```python
store_std=(
    df.groupby("Store")["Weekly_Sales"]
    .std()
    .sort_values() 
)

store_std.head(10)
```

```python
plt.figure(figsize=(12,6))

store_std.head(10).plot(
    kind="bar",
    color="green"
)

plt.title("Top 10 Most Consistent Stores")
plt.xlabel("Store")
plt.ylabel("Standard Deviation of Weekly Sales")

plt.xticks(rotation=0)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

# 14. Store-wise Sales Consistency

## Objective

To identify the most consistent Walmart stores by analyzing the standard deviation of weekly sales.

## Observation

- Store 37 has the lowest standard deviation, making it the most consistent store.
- Stores 30, 33, and 44 also show stable weekly sales.
- These stores have minimal week-to-week sales fluctuations.

## Business Insight

Stores with low sales variation are easier to forecast and manage. Businesses can use these stores as benchmarks for inventory planning and operational efficiency.
